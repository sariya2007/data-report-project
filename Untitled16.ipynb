```python
import numpy as np
import pandas as pd

# 1. Create a sample dataset for demonstration (or replace this with your own file upload)
np.random.seed(42)
data = {
    "CustomerID": range(1, 101),
    "Age": np.random.randint(18, 70, size=100),
    "Income": np.random.choice(
        [50000, 75000, 100000, np.nan, 120000], size=100
    ),
    "City": np.random.choice(["New York", "London", "Paris", "Tokyo"], size=100),
    "Department": np.random.choice(["HR", "Engineering", "Sales"], size=100),
}
df = pd.DataFrame(data)

# Introduce a couple of duplicate rows intentionally for testing
df = pd.concat([df, df.iloc[:2]], ignore_index=True)


# 2. Define the Report Function
def generate_data_report(df: pd.DataFrame):
  """Generates a comprehensive data overview report for a pandas DataFrame."""
  print("=" * 60)
  print("📊 INITIAL DATA REPORT")
  print("=" * 60)

  # 1. Data Shape
  print("\n--- 1. DATA SHAPE ---")
  rows, cols = df.shape
  print(f"Total Rows: {rows:,}")
  print(f"Total Columns: {cols:,}")

  # 2. Data Types & Memory Usage
  print("\n--- 2. DATA TYPES & MEMORY USAGE ---")
  print(f"Memory Usage: {df.memory_usage(deep=True).sum() / 1024 ** 2:.2f} MB")
  dtype_counts = df.dtypes.value_counts()
  for dtype, count in dtype_counts.items():
    print(f"  - {dtype}: {count} columns")

  # 3. Missing Values
  print("\n--- 3. MISSING VALUES ---")
  missing_count = df.isnull().sum()
  missing_pct = (missing_count / rows) * 100
  missing_df = pd.DataFrame(
      {"Missing Values": missing_count, "Percentage (%)": missing_pct}
  )
  missing_summary = missing_df[missing_df["Missing Values"] > 0]

  if missing_summary.empty:
    print("✅ No missing values found in the dataset.")
  else:
    print(missing_summary.to_string())

  # 4. Duplicate Rows
  print("\n--- 4. DUPLICATE VALUES ---")
  duplicates = df.duplicated().sum()
  dup_pct = (duplicates / rows) * 100
  print(f"Duplicate Rows: {duplicates:,} ({dup_pct:.2f}% of total rows)")

  # 5. Unique Values in Object / Categorical Columns
  print("\n--- 5. UNIQUE VALUES IN OBJECT COLUMNS ---")
  object_cols = df.select_dtypes(include=["object", "category"]).columns

  if len(object_cols) == 0:
    print("No object or categorical columns found.")
  else:
    for col in object_cols:
      unique_count = df[col].nunique()
      print(f"\nColumn: '{col}' (Unique Values: {unique_count})")
      if unique_count <= 10:
        print(f"  Values: {list(df[col].dropna().unique())}")
      else:
        print("  Top 5 most frequent values:")
        print(df[col].value_counts().head(5).to_string())

  print("\n" + "=" * 60)
  print("🏁 REPORT COMPLETE")
  print("=" * 60)


# 3. Run the Report
generate_data_report(df)
```
